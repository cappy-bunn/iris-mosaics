# Download IRIS level 1 data from JSOC

Queries JSOC for the FUV level 1 files covering a mosaic and downloads them
locally. This replaces the old `download_mosaic_lvl1.pro` IDL route, which was
retired because it did not retrieve all the files.

Timing: JSOC takes roughly **40 minutes** to process the export request (an email
arrives when it is ready); the download itself takes about **2.5 hours**.

Time ranges for previously processed mosaics are recorded in `docs/STATUS.md`.
Give the mosaic a 2-minute margin on either side.


```python
import sunpy.net
import sunpy.net.attrs
import pathlib as pl
```

Set the mosaic being downloaded.


```python
date_string = '20220507'
t_start = '2022-05-07 12:38:55'
t_end   = '2022-05-09 03:11:37'

notify_email = 'catharine.bunn@montana.edu'   # must be registered with JSOC
download_path = pl.Path(fr'D:\IRIS data\deep_mosaics\{date_string}\level_1')
```

Query JSOC. Inspect the response before fetching.


```python
response = sunpy.net.Fido.search(
    sunpy.net.attrs.Time(t_start, t_end),
    sunpy.net.attrs.jsoc.Series('iris.lev1'),
    sunpy.net.attrs.jsoc.Notify(notify_email),
    sunpy.net.attrs.jsoc.Keyword('INSTRUME') == "'FUV'",
)
response
```

Fetch. Expect ~40 min of JSOC processing before the transfer starts.


```python
%%time
download_path.mkdir(parents=True, exist_ok=True)
files = sunpy.net.Fido.fetch(response, path=str(download_path))
print(f'{len(files)} files downloaded to {download_path}')
```
